```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data = pd.read_csv('test/TelecomCustomersChurn.csv')
```

```python
#Supprimer les espaces inutiles
data['TotalCharges'] = data['TotalCharges'].str.strip()
```

```python
#convertir la colonne TotalCharges en numérique
data['TotalCharges'] = pd.to_numeric(data['TotalCharges'], errors='coerce')
```

```python
#Remplacer les valeurs manquantes par la médiane
data['TotalCharges'].fillna(data['TotalCharges'].median(), inplace=True)
```

```python
data.duplicated().sum()
```

```python
#Encoder les variables binaires en 0 et 1
data['Churn'] = data['Churn'].map({'Yes': 1, 'No': 0})
data['Partner'] = data['Partner'].map({'Yes': 1, 'No': 0})
data['Dependents'] = data['Dependents'].map({'Yes': 1, 'No': 0})
data['PhoneService'] = data['PhoneService'].map({'Yes': 1, 'No': 0})
data['PaperlessBilling'] = data['PaperlessBilling'].map({'Yes': 1, 'No': 0})
```

```python
# Normaliser les colonnes numériques
from sklearn.preprocessing import StandardScaler

# Sélectionner les colonnes numériques à normaliser
numerical_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

# Appliquer la normalisation
scaler = StandardScaler()
data[numerical_cols] = scaler.fit_transform(data[numerical_cols])

print("Données normalisées.")
```

```python
# Identifier les tendances réseau (analyse des charges mensuelles et tenure)
# Analyser les données de trafic pour optimiser les performances
network_trends = data[['tenure', 'MonthlyCharges']].rolling(window=7, min_periods=1).mean()
print(network_trends.head(10))
```

```python
# Définir les KPIs pertinents (QoS, taux de panne, anomalies)

# Convertir les colonnes de services en valeurs numériques
service_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']
for col in service_cols:
    data[col] = (data[col] == 'Yes').astype(int)
```

```python

# 1. QoS (Quality of Service)
qos_metrics = {
    'avg_tenure_normalized': data['tenure'].mean(),
    'avg_monthly_charges_normalized': data['MonthlyCharges'].mean(),
    'avg_services_per_customer': data[service_cols].sum(axis=1).mean(),
    'internet_service_fiber_pct': (data['InternetService'] == 'Fiber optic').sum() / len(data) * 100,
}
print("KPIs de QoS:")
```

```python

# 2. Taux de panne (Churn Rate)
churn_rate = (data['Churn'].sum() / len(data)) * 100
retained_customers = 100 - churn_rate
```

```python

# 3. Détection d'anomalies (IQR method) - sur les données normalisées
outliers = {}
for col in ['MonthlyCharges', 'TotalCharges']:
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    anomalies_count = ((data[col] < lower_bound) | (data[col] > upper_bound)).sum()
    outliers[col] = {'count': anomalies_count, 'pct': (anomalies_count / len(data)) * 100}
```

```python

# Affichage des KPIs
print("=" * 50)
print("TABLEAU DE BORD DES KPIs")
print("=" * 50)
print(f"\n1. QoS (Qualité de Service):")
for key, value in qos_metrics.items():
    print(f"   {key}: {value:.2f}")
print(f"\n2. Taux de Panne:")
print(f"   Churn Rate: {churn_rate:.2f}%")
print(f"   Taux de Rétention: {retained_customers:.2f}%")
print(f"\n3. Anomalies Détectées:")
for col, stats_dict in outliers.items():
    print(f"   {col}: {stats_dict['count']} anomalies ({stats_dict['pct']:.2f}%)")
```
